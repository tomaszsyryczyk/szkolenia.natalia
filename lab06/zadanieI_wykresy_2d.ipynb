# Zadanie I – Wykresy dwóch funkcji na jednym rysunku

## Funkcje do narysowania (przedział $[-2\pi, 2\pi]$)

$$f_0(x) = \begin{cases} 2\cos^2\!\left(\frac{x}{2}\right)\,|\sin(x)| & \text{jeśli } x \le 0 \\ 2\sin\!\left(\frac{x^2}{2}\right)\,|\cos(x)| & \text{jeśli } x > 0 \end{cases}$$

$$f_1(x) = \exp\!\left(\frac{\sin(x)}{x}\right) \cdot \exp(\cos(\pi x))$$

### Uwagi
- **f0** to funkcja sklejana – różna definicja po lewej i prawej stronie zera
- **f1** ma osobliwość pozorną w $x=0$: $\lim_{x\to 0}\frac{\sin x}{x}=1$, więc $f_1(0)=e^1\cdot e^1=e^2$

## Dlaczego używamy NumPy i Matplotlib?

### NumPy

NumPy (Numerical Python) to biblioteka do obliczeń numerycznych. W tym zadaniu korzystamy z niej z trzech powodów:

**1. Operacje na całych tablicach naraz (bez pętli `for`)**

Zwykły Python nie umie zastosować funkcji matematycznej do listy liczb w jednym kroku — trzeba pisać pętlę. NumPy pozwala pisać `np.sin(x)`, gdzie `x` to tablica 200 liczb, i od razu dostajemy 200 wyników. To jest szybsze i czytelniejsze:

```python
# Python bez NumPy – ręczna pętla:
wyniki = [math.sin(xi) for xi in x_lista]

# NumPy – operacja wektorowa (element-wise), działa na wszystkich elementach naraz:
wyniki = np.sin(x)
```

**2. Generowanie równomiernej siatki punktów osi X**

- `np.linspace(start, stop, n)` — tworzy dokładnie `n` punktów równomiernie rozłożonych w przedziale  
- `np.arange(start, stop, krok)` — tworzy punkty z zadanym krokiem (jak `range`, ale dla liczb zmiennoprzecinkowych)

Bez NumPy musielibyśmy sami liczyć każdy punkt i budować listę ręcznie.

**3. `np.where` — funkcja sklejana bez pętli**

Funkcja `f0` ma różną definicję dla x ≤ 0 i x > 0. `np.where(warunek, wartość_gdy_True, wartość_gdy_False)` wybiera element-wise który wynik zwrócić — bez żadnej pętli:

```python
# Bez NumPy:
wynik = [lewa(xi) if xi <= 0 else prawa(xi) for xi in x]

# Z NumPy – szybciej i czytelniej:
wynik = np.where(x <= 0, lewa(x), prawa(x))
```

---

### Matplotlib

Matplotlib to standardowa biblioteka do rysowania wykresów w Pythonie. Moduł `matplotlib.pyplot` (importowany jako `plt`) działa podobnie do środowiska MATLAB:

- `plt.subplots()` — tworzy okno wykresu i osie współrzędnych (`fig`, `ax`)
- `ax.plot(x, y)` — rysuje krzywą łączącą punkty
- Parametry `color`, `linestyle`, `linewidth` sterują wyglądem linii

Bez Matplotlib musielibyśmy korzystać z zewnętrznych narzędzi lub ręcznie obliczać piksele — biblioteka bierze to wszystko na siebie.

## Implementacja funkcji

````python
import numpy as np
import matplotlib.pyplot as plt

# --- f0: funkcja sklejana ---
def f0(x):
    """
    Piecewise: lewa gałąź (x<=0) i prawa gałąź (x>0).
    np.where działa element-wise na tablicach numpy.
    """
    lewy  = 2 * np.cos(x / 2)**2 * np.abs(np.sin(x))   # x <= 0
    prawy = 2 * np.sin(x**2 / 2) * np.abs(np.cos(x))   # x > 0
    return np.where(x <= 0, lewy, prawy)


# --- f1: exp(sinc) * exp(cos(pi*x)) ---
def f1(x):
    """
    W x=0 mamy sin(x)/x -> 1, więc osobliwość jest pozorna.
    np.where chroni przed dzieleniem przez 0.
    """
    sinc_x = np.where(x == 0, 1.0, np.sin(x) / x)   # sin(x)/x z limitem w 0
    return np.exp(sinc_x) * np.exp(np.cos(np.pi * x))
````

## Wariant 1 – linspace (200 punktów)

````python
# Generowanie równomiernie rozmieszczonych punktów
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

# Wykresy obu funkcji na jednym rysunku
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    x, f0(x),
    color='crimson',
    linestyle='-',       # linia ciągła
    linewidth=2.0,
    label=r'$f_0(x)$'
)
ax.plot(
    x, f1(x),
    color='seagreen',
    linestyle='--',      # linia przerywana – różni się od f0
    linewidth=1.2,
    label=r'$f_1(x)$'
)

ax.axhline(0, color='black', linewidth=0.5)  # oś X
ax.axvline(0, color='black', linewidth=0.5)  # oś Y
ax.set_title(r'Wykresy $f_0$ i $f_1$ na $[-2\pi, 2\pi]$ – linspace(200)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Wariant 2 – arange (step = 0.005)

````python
# np.arange z krokiem 0.005 – gęstszy zestaw punktów
x2 = np.arange(-2 * np.pi, 2 * np.pi, 0.005)
print(f"Liczba punktów w wariancie arange: {len(x2)}")

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    x2, f0(x2),
    color='royalblue',
    linestyle='-.',      # linia kropka-kreska
    linewidth=2.5,
    label=r'$f_0(x)$'
)
ax.plot(
    x2, f1(x2),
    color='darkorange',
    linestyle=':',       # linia kropkowana
    linewidth=1.0,
    label=r'$f_1(x)$'
)

ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.set_title(r'Wykresy $f_0$ i $f_1$ na $[-2\pi, 2\pi]$ – arange(step=0.005)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Weryfikacja wartości w x = 0

Sprawdzamy, że obsługa osobliwości f1(0) działa poprawnie.

````python
import math

# f0(0): obydwie gałęzie mają tę samą wartość w x=0
f0_at_0_lewy  = 2 * math.cos(0)**2 * abs(math.sin(0))  # = 0
f0_at_0_prawy = 2 * math.sin(0)    * abs(math.cos(0))  # = 0
print(f"f0(0) lewa gałąź  = {f0_at_0_lewy:.6f}")
print(f"f0(0) prawa gałąź = {f0_at_0_prawy:.6f}")
print(f"f0(0) numpy       = {f0(np.array([0.0]))[0]:.6f}")

# f1(0): limit sin(x)/x = 1
f1_limit = math.exp(1.0) * math.exp(math.cos(0))   # exp(1)*exp(1) = e^2
print(f"\nf1(0) z limitu    = {f1_limit:.6f}  (= e^2 = {math.e**2:.6f})")
print(f"f1(0) numpy       = {f1(np.array([0.0]))[0]:.6f}")
````
