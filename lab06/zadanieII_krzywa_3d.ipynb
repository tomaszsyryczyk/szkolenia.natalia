# Zadanie II – Wykres krzywej parametrycznej w R³

## Definicja krzywej

$$y: T \to \mathbb{R}^3, \quad t \in T = [0, t_{\max}]$$

$$y(t) = \bigl(x_0(t),\; x_1(t),\; x_2(t)\bigr)$$

gdzie:

$$x_0(t) = \sin(2\pi t)$$
$$x_1(t) = 2\cos(\pi t)$$
$$x_2(t) = t^2$$

## Parametry

$$t_{\max} = n \cdot dt = 400 \cdot 0.01 = 4.0$$

### Dwa warianty wykresu
1. **Scatter** – wykres punktowy (rozproszony), dowolny `color` i `marker`
2. **Line** – interpolacja odcinkami prostych, inny `color` i dowolna `linewidth`

## Generowanie danych

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D   # wymagane do wykresów 3D

# Parametry z treści zadania
n  = 400
dt = 0.01
t_max = n * dt         # = 4.0

# Tablica parametru t: n+1 punktów od 0 do t_max włącznie
t = np.arange(0, t_max + dt, dt)
print(f"t_max = {t_max}, punktów: {len(t)}, t[0]={t[0]}, t[-1]={t[-1]:.2f}")

# Składowe krzywej
x0 = np.sin(2 * np.pi * t)   # oscyluje w [-1, 1]
x1 = 2 * np.cos(np.pi * t)   # oscyluje w [-2, 2]
x2 = t**2                     # rośnie paraboliczne od 0 do t_max^2
```

## Wykres 1 – Scatter (punktowy)

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# scatter: każdy punkt osobno, bez połączeń
# marker='.' – małe kropki; s=2 – rozmiar markera
ax.scatter(
    x0, x1, x2,
    color='royalblue',
    marker='.',
    s=2,
    label='y(t) – scatter'
)

ax.set_title('Krzywa parametryczna y(t) w R³ – wykres punktowy')
ax.set_xlabel(r'$x_0 = \sin(2\pi t)$')
ax.set_ylabel(r'$x_1 = 2\cos(\pi t)$')
ax.set_zlabel(r'$x_2 = t^2$')
ax.legend()

plt.tight_layout()
plt.show()
```

## Wykres 2 – Line (interpolacja odcinkami prostych)

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# plot: kolejne punkty połączone odcinkami prostymi
ax.plot(
    x0, x1, x2,
    color='deeppink',
    linewidth=1.5,
    label='y(t) – line'
)

ax.set_title('Krzywa parametryczna y(t) w R³ – interpolacja liniowa')
ax.set_xlabel(r'$x_0 = \sin(2\pi t)$')
ax.set_ylabel(r'$x_1 = 2\cos(\pi t)$')
ax.set_zlabel(r'$x_2 = t^2$')
ax.legend()

plt.tight_layout()
plt.show()
```

## Oba wykresy obok siebie (porównanie)

```python
fig = plt.figure(figsize=(14, 6))

# Lewy subplot – scatter
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(x0, x1, x2, color='royalblue', marker='.', s=2)
ax1.set_title('Scatter')
ax1.set_xlabel(r'$x_0$')
ax1.set_ylabel(r'$x_1$')
ax1.set_zlabel(r'$x_2$')

# Prawy subplot – line
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot(x0, x1, x2, color='deeppink', linewidth=1.5)
ax2.set_title('Line')
ax2.set_xlabel(r'$x_0$')
ax2.set_ylabel(r'$x_1$')
ax2.set_zlabel(r'$x_2$')

plt.suptitle(r'Krzywa $y(t) = (\sin 2\pi t,\; 2\cos\pi t,\; t^2)$, $t\in[0,4]$')
plt.tight_layout()
plt.show()
```

## Interpretacja kształtu krzywej

| Składowa | Opis |
|---|---|
| $x_0(t) = \sin(2\pi t)$ | oscyluje z częstotliwością 2 (2 pełne cykle na jednostkę t) |
| $x_1(t) = 2\cos(\pi t)$ | oscyluje z częstotliwością 1, amplituda 2 |
| $x_2(t) = t^2$ | rośnie monotonicznie – oś "głębokości" wykresu |

Krzywa tworzy **spiralę** owijającą się wokół rosnącej osi $x_2$:  
elipsy w płaszczyźnie $(x_0, x_1)$ przesuwają się coraz wyżej wzdłuż $x_2 = t^2$.
