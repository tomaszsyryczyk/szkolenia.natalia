# Zadanie 1 – Rekurencyjne wyszukiwanie binarne

## Czego wymaga zadanie?

Zaimplementować rekurencyjną funkcję `binary_rek(a, first, last, x)`, która:  
- przeszukuje **posortowaną** listę `a`  
- zwraca **indeks** `i` taki, że `a[i] == x`  
- zwraca **-1**, jeśli `x` nie istnieje w liście  

Wywołanie startowe: `binary_rek(a, 0, len(a)-1, x)`

## Idea algorytmu

Wyszukiwanie binarne działa jak szukanie słowa w słowniku:  
otwieramy na środku → jeśli szukane słowo jest wcześniej alfabetycznie, odrzucamy prawą połowę; jeśli później – odrzucamy lewą.  
Powtarzamy na coraz mniejszym fragmencie, aż znajdziemy lub stwierdzimy brak.

**Dlaczego rekurencja?**  
Każde wywołanie rozwiązuje ten sam problem, ale na **połowie** danych – naturalny podział na podproblemy.

## Algorytm krok po kroku

1. Oblicz `middle = (first + last) // 2` (dzielenie całkowitoliczbowe `//`)
2. Jeśli `first > last` → zwróć `-1` (brak elementu, przedział pusty)
3. Jeśli `x == a[middle]` → znaleziono, zwróć `middle`
4. Jeśli `x < a[middle]` → szukaj w lewej połowie: `binary_rek(a, first, middle-1, x)`
5. W przeciwnym razie → szukaj w prawej połowie: `binary_rek(a, middle+1, last, x)`

## Implementacja

````python
def binary_rek(a, first, last, x):
    """
    Rekurencyjne wyszukiwanie binarne elementu x w posortowanej liście a.
    Zwraca indeks elementu lub -1 jeśli nie znaleziono.
    """
    # Warunek bazowy: przedział pusty – element nie istnieje w liście
    if first > last:
        return -1

    # Indeks środkowy – MUSI być dzielenie całkowitoliczbowe //
    # Zwykłe / dałoby float, np. 2.5, co nie może być indeksem listy
    middle = (first + last) // 2

    # Warunek bazowy: znaleziono element
    if x == a[middle]:
        return middle

    # Rekurencja w lewą stronę: x jest mniejszy niż element środkowy
    # Odrzucamy prawą połowę – new last = middle - 1
    elif x < a[middle]:
        return binary_rek(a, first, middle - 1, x)

    # Rekurencja w prawą stronę: x jest większy niż element środkowy
    # Odrzucamy lewą połowę – new first = middle + 1
    else:
        return binary_rek(a, middle + 1, last, x)
````

### Wizualizacja działania – drzewo wywołań dla przykładu

Lista: `[1, 3, 5, 7, 9, 11, 13]`, szukamy `x = 9`

```
binary_rek(a, 0, 6, 9)  →  middle=3, a[3]=7, 9 > 7 → prawa połowa
  binary_rek(a, 4, 6, 9)  →  middle=5, a[5]=11, 9 < 11 → lewa połowa
    binary_rek(a, 4, 4, 9)  →  middle=4, a[4]=9, ZNALEZIONO → zwróć 4
```

3 wywołania zamiast 5 przy liniowym przeszukiwaniu.

## Testy

````python
# Lista testowa – musi być posortowana
a = [1, 3, 5, 7, 9, 11, 13, 17, 21, 25]
n = len(a)
print(f"Lista: {a}")
print(f"Długość: {n}\n")

# Testy wyszukiwania
test_cases = [
    (1,  "pierwszy element"),
    (25, "ostatni element"),
    (7,  "element środkowy"),
    (9,  "element w prawej połowie"),
    (4,  "element nieistniejący (parzyste)"),
    (0,  "element mniejszy niż minimum"),
    (99, "element większy niż maksimum"),
]

print(f"{'x':>4}  {'wynik':>6}  {'opis'}")
print("-" * 45)
for x, opis in test_cases:
    idx = binary_rek(a, 0, n - 1, x)
    if idx == -1:
        info = "nie znaleziono"
    else:
        info = f"a[{idx}] = {a[idx]}"
    print(f"{x:>4}  {idx:>6}  {info}  ({opis})")
````

````python
# Automatyczna weryfikacja: sprawdź wszystkie elementy listy
print("Weryfikacja: szukamy każdego elementu listy (powinny zwrócić poprawny indeks):")
bledy = 0
for i, val in enumerate(a):
    wynik = binary_rek(a, 0, n - 1, val)
    if wynik != i:
        print(f"  BLAD: x={val}, oczekiwany indeks={i}, otrzymano={wynik}")
        bledy += 1

if bledy == 0:
    print("  Wszystkie elementy znalezione poprawnie!")

# Weryfikacja braku elementów
print("\nWeryfikacja: szukamy elementów spoza listy (powinny zwrócić -1):")
brak = [0, 2, 4, 6, 8, 10, 12, 100]
bledy2 = 0
for val in brak:
    wynik = binary_rek(a, 0, n - 1, val)
    if wynik != -1:
        print(f"  BLAD: x={val} powinno zwrócić -1, zwróciło {wynik}")
        bledy2 += 1

if bledy2 == 0:
    print("  Wszystkie braki wykryte poprawnie!")
````

## Złożoność algorytmu

Przy każdym wywołaniu rekurencyjnym przedział **zmniejsza się o połowę**.  
Liczba wywołań to $\lceil \log_2 n \rceil$, więc złożoność to $O(\log n)$.

Dla listy `n = 1 000 000` elementów wystarczy **20 porównań** (zamiast miliona przy wyszukiwaniu liniowym).

````python
import math

# Licznik wywołań rekurencyjnych
licznik = [0]

def binary_rek_z_licznikiem(a, first, last, x):
    licznik[0] += 1
    if first > last:
        return -1
    middle = (first + last) // 2
    if x == a[middle]:
        return middle
    elif x < a[middle]:
        return binary_rek_z_licznikiem(a, first, middle - 1, x)
    else:
        return binary_rek_z_licznikiem(a, middle + 1, last, x)

print(f"{'n':>8} {'log2(n)':>10} {'wywołań':>10}")
print("-" * 32)
for exp in range(1, 15):
    n = 2 ** exp
    lista = list(range(0, 2 * n, 2))  # posortowana lista parzystych
    licznik[0] = 0
    binary_rek_z_licznikiem(lista, 0, n - 1, -1)  # szukamy czegoś nieistniejącego – worst case
    print(f"{n:>8} {math.log2(n):>10.1f} {licznik[0]:>10}")
````

## Podsumowanie

| Element | Opis |
|---|---|
| **Funkcja** | `binary_rek(a, first, last, x)` |
| **Warunek bazowy 1** | `first > last` → zwraca `-1` (brak elementu) |
| **Warunek bazowy 2** | `x == a[middle]` → zwraca `middle` (znaleziono) |
| **Rekurencja lewa** | `x < a[middle]` → `binary_rek(a, first, middle-1, x)` |
| **Rekurencja prawa** | `x > a[middle]` → `binary_rek(a, middle+1, last, x)` |
| **Kluczowy detal** | `middle = (first + last) // 2` — obowiązkowo `//` |
| **Złożoność** | $O(\log n)$ — logarytmiczna |
